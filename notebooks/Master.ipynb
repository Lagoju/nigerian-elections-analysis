```python
#11
def standardize_state_names(df):
    df = df.copy()

    df["state"] = (
        df["state"]
        .astype(str)
        .str.strip()
        .str.title()
    )

    # Preserve the conventional FCT abbreviation
    df["state"] = df["state"].replace({
        "Fct": "FCT"
    })

    return df


df_2015 = standardize_state_names(df_2015)
df_2019 = standardize_state_names(df_2019)
df_2023 = standardize_state_names(df_2023)
```

```python
#35
def get_comparable_winner(row):
    party_votes = {
        "APC": row["apc_votes"],
        "PDP": row["pdp_votes"],
        "Other": row["other_party_votes"]
    }

    return max(party_votes, key=party_votes.get)


for df in [df_2015, df_2019, df_2023]:
    df["comparable_winner_party"] = df.apply(
        get_comparable_winner,
        axis=1
    )
```

```python
#36
def get_winner_votes(row):
    party_votes = {
        "APC": row["apc_votes"],
        "PDP": row["pdp_votes"],
        "Other": row["other_party_votes"]
    }

    return max(party_votes.values())


for df in [df_2015, df_2019, df_2023]:
    df["comparable_winner_votes"] = df.apply(
        get_winner_votes,
        axis=1
    )
```

```python
#37
def get_runner_up_votes(row):
    party_votes = sorted(
        [
            row["apc_votes"],
            row["pdp_votes"],
            row["other_party_votes"]
        ],
        reverse=True
    )

    return party_votes[1]


for df in [df_2015, df_2019, df_2023]:
    df["comparable_runner_up_votes"] = df.apply(
        get_runner_up_votes,
        axis=1
    )
```

```python
#38
for df in [df_2015, df_2019, df_2023]:

    df["comparable_margin_votes"] = (
        df["comparable_winner_votes"]
        - df["comparable_runner_up_votes"]
    )
```

```python
#39
for df in [df_2015, df_2019, df_2023]:

    df["comparable_winner_vote_share"] = (
        df["comparable_winner_votes"]
        / df["valid_votes"]
        * 100
    )
```

```python
#40
for df in [df_2015, df_2019, df_2023]:

    df["comparable_margin_percentage_points"] = (
        df["comparable_margin_votes"]
        / df["valid_votes"]
        * 100
    )
```

```python
#41---EFFECTIVE NUMBER OF PARTIES
for df in [df_2015, df_2019, df_2023]:

    apc_p = df["apc_votes"] / df["valid_votes"]
    pdp_p = df["pdp_votes"] / df["valid_votes"]
    other_p = df["other_party_votes"] / df["valid_votes"]

    df["enp_comparable"] = (
        1 / (
            apc_p**2
            + pdp_p**2
            + other_p**2
        )
    )
```

```python
# CELL 102 — Calculate national participation and vote-quality metrics

national_summary["turnout_rate"] = (
    national_summary["accredited_voters"]
    / national_summary["registered_voters"]
    * 100
)

national_summary["accreditation_rate"] = (
    national_summary["accredited_voters"]
    / national_summary["registered_voters"]
    * 100
)

national_summary["rejection_rate"] = (
    national_summary["rejected_votes"]
    / national_summary["total_votes_cast"]
    * 100
)

national_summary["valid_vote_rate"] = (
    national_summary["valid_votes"]
    / national_summary["total_votes_cast"]
    * 100
)

national_summary
```

```python
# CELL 103 — Calculate national comparable party vote shares

national_summary["apc_vote_share"] = (
    national_summary["apc_votes"]
    / national_summary["valid_votes"]
    * 100
)

national_summary["pdp_vote_share"] = (
    national_summary["pdp_votes"]
    / national_summary["valid_votes"]
    * 100
)

national_summary["other_party_vote_share"] = (
    national_summary["other_party_votes"]
    / national_summary["valid_votes"]
    * 100
)

national_summary
```

```python
# CELL 104 — Calculate national comparable winner and electoral margin

def get_national_winner(row):
    party_votes = {
        "APC": row["apc_votes"],
        "PDP": row["pdp_votes"],
        "Other": row["other_party_votes"]
    }
    
    sorted_parties = sorted(
        party_votes.items(),
        key=lambda x: x[1],
        reverse=True
    )
    
    winner = sorted_parties[0]
    runner_up = sorted_parties[1]
    
    return pd.Series({
        "comparable_winner_party": winner[0],
        "comparable_winner_votes": winner[1],
        "comparable_runner_up_votes": runner_up[1],
        "comparable_margin_votes": winner[1] - runner_up[1],
        "comparable_winner_vote_share": (
            winner[1] / row["valid_votes"] * 100
        ),
        "comparable_margin_percentage_points": (
            (winner[1] - runner_up[1])
            / row["valid_votes"]
            * 100
        )
    })


national_winner_metrics = national_summary.apply(
    get_national_winner,
    axis=1
)

national_summary = pd.concat(
    [national_summary, national_winner_metrics],
    axis=1
)

national_summary
```

```python
# CELL 105 — Calculate national effective number of parties under the common structure

national_summary["enp_comparable"] = (
    1 / (
        (national_summary["apc_votes"] / national_summary["valid_votes"]) ** 2
        + (national_summary["pdp_votes"] / national_summary["valid_votes"]) ** 2
        + (national_summary["other_party_votes"] / national_summary["valid_votes"]) ** 2
    )
)

national_summary
```

```python
# CELL 135 — Calculate absolute changes in party vote shares

volatility_df["abs_apc_change"] = (
    volatility_df["apc_vote_share_change_pp"].abs()
)

volatility_df["abs_pdp_change"] = (
    volatility_df["pdp_vote_share_change_pp"].abs()
)

volatility_df["abs_other_change"] = (
    volatility_df["other_vote_share_change_pp"].abs()
)

volatility_df.head()
```

```python
# CELL 136 — Calculate Pedersen-style electoral volatility by state

volatility_df["electoral_volatility"] = (
    volatility_df[
        [
            "abs_apc_change",
            "abs_pdp_change",
            "abs_other_change"
        ]
    ].sum(axis=1)
    / 2
)

volatility_df.head()
```

```python
# CELL 138 — Summarize electoral volatility across election periods

volatility_summary = (
    volatility_df
    .groupby("election_transition")
    .agg(
        mean_volatility=("electoral_volatility", "mean"),
        median_volatility=("electoral_volatility", "median"),
        minimum_volatility=("electoral_volatility", "min"),
        maximum_volatility=("electoral_volatility", "max"),
        standard_deviation=("electoral_volatility", "std")
    )
    .reset_index()
)

volatility_summary.round(2)
```

```python
# CELL 141 — Identify which party experienced the largest vote-share movement

def identify_largest_change(row):
    changes = {
        "APC": row["abs_apc_change"],
        "PDP": row["abs_pdp_change"],
        "Other": row["abs_other_change"]
    }
    
    return max(changes, key=changes.get)


volatility_df["largest_party_movement"] = (
    volatility_df.apply(
        identify_largest_change,
        axis=1
    )
)

volatility_df[
    [
        "state",
        "election_transition",
        "electoral_volatility",
        "largest_party_movement"
    ]
].head()
```

```python
# CELL 154 — Calculate APC, PDP and Other vote shares by geopolitical zone

zone_party_summary = (
    state_analysis
    .groupby(
        ["election_year", "geopolitical_zone"]
    )
    .agg(
        apc_votes=("apc_votes", "sum"),
        pdp_votes=("pdp_votes", "sum"),
        other_party_votes=("other_party_votes", "sum"),
        valid_votes=("valid_votes", "sum")
    )
    .reset_index()
)

zone_party_summary["apc_vote_share"] = (
    zone_party_summary["apc_votes"]
    / zone_party_summary["valid_votes"]
    * 100
)

zone_party_summary["pdp_vote_share"] = (
    zone_party_summary["pdp_votes"]
    / zone_party_summary["valid_votes"]
    * 100
)

zone_party_summary["other_vote_share"] = (
    zone_party_summary["other_party_votes"]
    / zone_party_summary["valid_votes"]
    * 100
)

zone_party_summary.round(2)
```

```python
# CELL 155 — Display 2023 party support by geopolitical zone

zone_2023 = (
    zone_party_summary[
        zone_party_summary["election_year"] == 2023
    ]
    [
        [
            "geopolitical_zone",
            "apc_vote_share",
            "pdp_vote_share",
            "other_vote_share"
        ]
    ]
    .sort_values(
        "other_vote_share",
        ascending=False
    )
)

zone_2023.round(2)
```

```python
# CELL 156 — Calculate full-party 2023 vote shares by geopolitical zone

zone_2023_full = (
    state_analysis[
        state_analysis["election_year"] == 2023
    ]
    .groupby("geopolitical_zone")
    .agg(
        apc_votes=("apc_votes", "sum"),
        pdp_votes=("pdp_votes", "sum"),
        lp_votes=("lp_votes", "sum"),
        nnpp_votes=("nnpp_votes", "sum"),
        other_votes=("other_2023_votes", "sum"),
        valid_votes=("valid_votes", "sum")
    )
    .reset_index()
)

for party in [
    "apc",
    "pdp",
    "lp",
    "nnpp",
    "other"
]:
    zone_2023_full[f"{party}_vote_share"] = (
        zone_2023_full[f"{party}_votes"]
        / zone_2023_full["valid_votes"]
        * 100
    )

zone_2023_full.round(2)
```

```python
# CELL 157 — Identify the geopolitical zone with the highest vote share for each major 2023 party

major_2023_parties = [
    "apc",
    "pdp",
    "lp",
    "nnpp"
]

for party in major_2023_parties:
    row = zone_2023_full.loc[
        zone_2023_full[f"{party}_vote_share"].idxmax()
    ]
    
    print(
        f"{party.upper()}: "
        f"{row['geopolitical_zone']} "
        f"({row[f'{party}_vote_share']:.2f}%)"
    )
```

```python
# CELL 158 — Calculate turnout by geopolitical zone

zone_turnout = (
    state_analysis
    .groupby(
        ["election_year", "geopolitical_zone"]
    )
    .agg(
        registered_voters=("registered_voters", "sum"),
        accredited_voters=("accredited_voters", "sum")
    )
    .reset_index()
)

zone_turnout["turnout_rate"] = (
    zone_turnout["accredited_voters"]
    / zone_turnout["registered_voters"]
    * 100
)

zone_turnout.round(2)
```

```python
# CELL 159 — Calculate turnout changes by geopolitical zone

zone_turnout = zone_turnout.sort_values(
    ["geopolitical_zone", "election_year"]
)

zone_turnout["previous_turnout_rate"] = (
    zone_turnout
    .groupby("geopolitical_zone")["turnout_rate"]
    .shift(1)
)

zone_turnout["turnout_change_pp"] = (
    zone_turnout["turnout_rate"]
    - zone_turnout["previous_turnout_rate"]
)

zone_turnout[
    zone_turnout["election_year"] != 2015
].round(2)
```

```python
# CELL 160 — Calculate geopolitical-zone party-share changes

zone_party_changes = zone_party_summary.sort_values(
    ["geopolitical_zone", "election_year"]
).copy()

for column in [
    "apc_vote_share",
    "pdp_vote_share",
    "other_vote_share"
]:
    zone_party_changes[f"{column}_change_pp"] = (
        zone_party_changes
        .groupby("geopolitical_zone")[column]
        .diff()
    )

zone_party_changes[
    zone_party_changes["election_year"] != 2015
].round(2)
```

```python
# CELL 194 — Correct polling-unit coverage classification and counts

# Rebuild the category classifier using the actual ZIP naming convention.
# The files use suffixes such as:
#   STATE_crosschecked.csv
#   STATE_unsure.csv
#   STATE_notfound.csv


# Import libraries required by this cell.
import os
import re
import zipfile
from collections import defaultdict

def identify_category(path):
    filename = os.path.basename(path).lower()
    
    if "_crosschecked" in filename:
        return "crosschecked"
    
    elif "_unsure" in filename:
        return "unsure"
    
    elif "_notfound" in filename:
        return "not_found"
    
    else:
        return "other"


# Nigerian states and FCT
nigeria_states = [
    "Abia", "Adamawa", "Akwa Ibom", "Anambra", "Bauchi",
    "Bayelsa", "Benue", "Borno", "Cross River", "Delta",
    "Ebonyi", "Edo", "Ekiti", "Enugu", "FCT", "Gombe",
    "Imo", "Jigawa", "Kaduna", "Kano", "Katsina", "Kebbi",
    "Kogi", "Kwara", "Lagos", "Nasarawa", "Niger", "Ogun",
    "Ondo", "Osun", "Oyo", "Plateau", "Rivers", "Sokoto",
    "Taraba", "Yobe", "Zamfara"
]


# Extract state directly from the filename
def extract_state_from_filename(filename):
    
    filename_clean = os.path.splitext(
        os.path.basename(filename)
    )[0]
    
    # Remove category suffixes
    filename_clean = re.sub(
        r"_(crosschecked|unsure|notfound)$",
        "",
        filename_clean,
        flags=re.IGNORECASE
    )
    
    # Match against known states
    for state in sorted(
        nigeria_states,
        key=len,
        reverse=True
    ):
        if filename_clean.lower() == state.lower():
            return state
    
    return None


# Rebuild coverage counts from the ZIP itself
coverage_records = defaultdict(
    lambda: {
        "crosschecked": 0,
        "unsure": 0,
        "not_found": 0
    }
)

unclassified_files = []

with zipfile.ZipFile(zip_path, "r") as z:
    
    for file in all_csv_files:
        
        category = identify_category(file)
        
        if category == "other":
            unclassified_files.append(file)
            continue
        
        state = extract_state_from_filename(file)
        
        if state is None:
            unclassified_files.append(file)
            continue
        
        # Count data rows without assuming anything about the CSV schema
        with z.open(file) as f:
            row_count = sum(1 for _ in f) - 1
        
        coverage_records[state][category] += row_count


# Convert to dataframe
coverage_by_state = (
    pd.DataFrame
    .from_dict(
        coverage_records,
        orient="index"
    )
    .reset_index()
    .rename(columns={"index": "state"})
)


# Ensure all categories are numeric
for col in [
    "crosschecked",
    "unsure",
    "not_found"
]:
    coverage_by_state[col] = (
        pd.to_numeric(
            coverage_by_state[col],
            errors="coerce"
        )
        .fillna(0)
        .astype(int)
    )


# Calculate total records
coverage_by_state["all_records"] = (
    coverage_by_state["crosschecked"]
    + coverage_by_state["unsure"]
    + coverage_by_state["not_found"]
)


# Calculate coverage percentages
coverage_by_state["crosschecked_share"] = (
    coverage_by_state["crosschecked"]
    / coverage_by_state["all_records"]
    * 100
)

coverage_by_state["uncertain_share"] = (
    (
        coverage_by_state["unsure"]
        + coverage_by_state["not_found"]
    )
    / coverage_by_state["all_records"]
    * 100
)


# Display validation information
print(
    "Jurisdictions identified:",
    len(coverage_by_state)
)

print(
    "Total crosschecked records:",
    f"{coverage_by_state['crosschecked'].sum():,}"
)

print(
    "Total unsure records:",
    f"{coverage_by_state['unsure'].sum():,}"
)

print(
    "Total not-found records:",
    f"{coverage_by_state['not_found'].sum():,}"
)

print(
    "Unclassified CSV files:",
    len(unclassified_files)
)

if unclassified_files:
    print("\nUnclassified files:")
    for file in unclassified_files:
        print(file)

display(
    coverage_by_state
    .sort_values("crosschecked_share")
)
```

```python
# CELL 202 — Build the 2023 polling-unit analysis dataset

# Start from the crosschecked polling-unit dataset.
pu_analysis = polling_crosschecked.copy()


# Ensure core vote columns are numeric.
vote_columns = [
    "apc_votes",
    "lp_votes",
    "pdp_votes",
    "nnpp_votes",
    "accredited_voters",
    "registered_voters"
]

for col in vote_columns:
    pu_analysis[col] = pd.to_numeric(
        pu_analysis[col],
        errors="coerce"
    )


# Calculate the combined votes recorded for the four major parties
# available in this polling-unit source.
pu_analysis["four_party_votes"] = (
    pu_analysis["apc_votes"]
    + pu_analysis["lp_votes"]
    + pu_analysis["pdp_votes"]
    + pu_analysis["nnpp_votes"]
)


# Calculate each party's share of the four-party vote total.
for party in ["apc", "lp", "pdp", "nnpp"]:
    
    pu_analysis[f"{party}_vote_share"] = (
        pu_analysis[f"{party}_votes"]
        / pu_analysis["four_party_votes"]
        * 100
    )


# Calculate accredited voters as a turnout proxy.
# This is NOT official turnout because the source does not provide
# complete total votes cast at polling-unit level.
pu_analysis["accreditation_rate"] = (
    pu_analysis["accredited_voters"]
    / pu_analysis["registered_voters"]
    * 100
)


# Identify the polling-unit winner among APC, LP, PDP and NNPP.
party_vote_columns = {
    "APC": "apc_votes",
    "LP": "lp_votes",
    "PDP": "pdp_votes",
    "NNPP": "nnpp_votes"
}

pu_analysis["polling_unit_winner"] = (
    pu_analysis[
        list(party_vote_columns.values())
    ]
    .idxmax(axis=1)
    .map({
        value: key
        for key, value in party_vote_columns.items()
    })
)


# Calculate winner and runner-up votes.
sorted_party_votes = (
    pu_analysis[
        list(party_vote_columns.values())
    ]
    .apply(
        lambda row: sorted(
            row.dropna(),
            reverse=True
        ),
        axis=1
    )
)

pu_analysis["winner_votes"] = sorted_party_votes.str[0]
pu_analysis["runner_up_votes"] = sorted_party_votes.str[1]


# Calculate the winning margin in votes.
pu_analysis["winner_margin_votes"] = (
    pu_analysis["winner_votes"]
    - pu_analysis["runner_up_votes"]
)


# Calculate winner's share of the four-party vote.
pu_analysis["winner_vote_share"] = (
    pu_analysis["winner_votes"]
    / pu_analysis["four_party_votes"]
    * 100
)


# Flag polling units where the four-party vote total exceeds
# the number of accredited voters.
pu_analysis["votes_exceed_accredited"] = (
    pu_analysis["four_party_votes"]
    > pu_analysis["accredited_voters"]
)


print(
    "Polling-unit observations:",
    f"{len(pu_analysis):,}"
)

print(
    "States:",
    pu_analysis["state"].nunique()
)

print(
    "Winner missing:",
    pu_analysis["polling_unit_winner"].isna().sum()
)

print(
    "Vote total > accredited:",
    pu_analysis["votes_exceed_accredited"].sum()
)
```

```python
# CELL 203 — Calculate national polling-unit winner distribution

# Count polling units won by each party.
pu_winner_summary = (
    pu_analysis["polling_unit_winner"]
    .value_counts()
    .rename_axis("party")
    .reset_index(name="polling_units_won")
)


# Calculate each party's share of crosschecked polling units won.
pu_winner_summary["share_of_crosschecked_units"] = (
    pu_winner_summary["polling_units_won"]
    / len(pu_analysis)
    * 100
)


# Display the national polling-unit winner distribution.
display(
    pu_winner_summary
)
```

```python
# CELL 204 — Calculate polling-unit winners by state

# Count polling-unit wins for each party within each state.
pu_winners_by_state = (
    pd.crosstab(
        pu_analysis["state"],
        pu_analysis["polling_unit_winner"]
    )
    .reset_index()
)


# Ensure all four major parties are represented as columns.
for party in ["APC", "LP", "PDP", "NNPP"]:
    if party not in pu_winners_by_state.columns:
        pu_winners_by_state[party] = 0


# Calculate total crosschecked polling units represented in each state.
pu_winners_by_state["total_crosschecked_units"] = (
    pu_winners_by_state[
        ["APC", "LP", "PDP", "NNPP"]
    ]
    .sum(axis=1)
)


# Calculate each party's share of polling-unit wins within the state.
for party in ["APC", "LP", "PDP", "NNPP"]:
    
    pu_winners_by_state[f"{party}_win_share"] = (
        pu_winners_by_state[party]
        / pu_winners_by_state["total_crosschecked_units"]
        * 100
    )


# Sort states by LP's polling-unit win share.
pu_winners_by_state = (
    pu_winners_by_state
    .sort_values("LP_win_share", ascending=False)
    .reset_index(drop=True)
)


display(
    pu_winners_by_state[
        [
            "state",
            "total_crosschecked_units",
            "APC",
            "LP",
            "PDP",
            "NNPP",
            "APC_win_share",
            "LP_win_share",
            "PDP_win_share",
            "NNPP_win_share"
        ]
    ]
)
```

```python
# CELL 205 — Identify the dominant polling-unit winner by state

# Identify which party won the largest number of crosschecked polling units
# in each state.

winner_columns = ["APC", "LP", "PDP", "NNPP"]

pu_winners_by_state["dominant_pu_winner"] = (
    pu_winners_by_state[winner_columns]
    .idxmax(axis=1)
)

pu_winners_by_state["dominant_pu_wins"] = (
    pu_winners_by_state[winner_columns]
    .max(axis=1)
)

pu_winners_by_state["dominant_pu_win_share"] = (
    pu_winners_by_state["dominant_pu_wins"]
    / pu_winners_by_state["total_crosschecked_units"]
    * 100
)

display(
    pu_winners_by_state[
        [
            "state",
            "total_crosschecked_units",
            "dominant_pu_winner",
            "dominant_pu_wins",
            "dominant_pu_win_share"
        ]
    ]
    .sort_values("dominant_pu_win_share", ascending=False)
)
```

```python
# CELL 206 — Calculate polling-unit party vote totals by state

# Aggregate the four major party votes across crosschecked polling units.

pu_state_votes = (
    pu_analysis
    .groupby("state", as_index=False)
    .agg(
        apc_votes=("apc_votes", "sum"),
        lp_votes=("lp_votes", "sum"),
        pdp_votes=("pdp_votes", "sum"),
        nnpp_votes=("nnpp_votes", "sum"),
        accredited_voters=("accredited_voters", "sum"),
        registered_voters=("registered_voters", "sum"),
        polling_units=("pu_code", "count")
    )
)

pu_state_votes["four_party_votes"] = (
    pu_state_votes["apc_votes"]
    + pu_state_votes["lp_votes"]
    + pu_state_votes["pdp_votes"]
    + pu_state_votes["nnpp_votes"]
)

for party in ["apc", "lp", "pdp", "nnpp"]:
    pu_state_votes[f"{party}_vote_share"] = (
        pu_state_votes[f"{party}_votes"]
        / pu_state_votes["four_party_votes"]
        * 100
    )

display(
    pu_state_votes.sort_values(
        "four_party_votes",
        ascending=False
    )
)
```

```python
# CELL 208 — Identify the leading party by aggregated polling-unit votes

party_share_columns = [
    "apc_vote_share",
    "lp_vote_share",
    "pdp_vote_share",
    "nnpp_vote_share"
]

party_name_map = {
    "apc_vote_share": "APC",
    "lp_vote_share": "LP",
    "pdp_vote_share": "PDP",
    "nnpp_vote_share": "NNPP"
}

pu_state_votes["vote_share_leader"] = (
    pu_state_votes[party_share_columns]
    .idxmax(axis=1)
    .map(party_name_map)
)

pu_state_votes["leading_vote_share"] = (
    pu_state_votes[party_share_columns]
    .max(axis=1)
)

display(
    pu_state_votes[
        [
            "state",
            "vote_share_leader",
            "leading_vote_share",
            "apc_vote_share",
            "lp_vote_share",
            "pdp_vote_share",
            "nnpp_vote_share"
        ]
    ]
    .sort_values("leading_vote_share", ascending=False)
)
```

```python
# CELL 211 — Calculate polling-unit vote concentration

# Calculate the Herfindahl-style concentration index across
# APC, LP, PDP and NNPP at each polling unit.

party_proportions = (
    pu_analysis[
        [
            "apc_vote_share",
            "lp_vote_share",
            "pdp_vote_share",
            "nnpp_vote_share"
        ]
    ]
    / 100
)

pu_analysis["vote_concentration"] = (
    party_proportions.pow(2).sum(axis=1)
)

display(
    pu_analysis[
        "vote_concentration"
    ].describe()
)
```

```python
# CELL 213 — Compare polling-unit winners with official 2023 state winners

# Standardize polling-unit state names so they match the longitudinal
# analysis dataset. The polling-unit source uses uppercase names.

pu_winner_comparison = pu_winners_by_state.copy()

pu_winner_comparison["state"] = (
    pu_winner_comparison["state"]
    .astype(str)
    .str.strip()
    .str.title()
)

pu_winner_comparison["state"] = (
    pu_winner_comparison["state"]
    .replace({"Fct": "FCT"})
)


# Extract official full-party 2023 state winners.
official_state_winners = (
    analysis_df[
        analysis_df["election_year"] == 2023
    ][
        [
            "state",
            "winner_2023_full_party"
        ]
    ]
    .drop_duplicates("state")
)


# Compare polling-unit dominant winner with official winner.
state_winner_comparison = (
    pu_winner_comparison[
        [
            "state",
            "dominant_pu_winner",
            "dominant_pu_win_share"
        ]
    ]
    .merge(
        official_state_winners,
        on="state",
        how="left",
        validate="one_to_one"
    )
)


# Check whether the two winner classifications agree.
state_winner_comparison["winner_agreement"] = (
    state_winner_comparison["dominant_pu_winner"]
    == state_winner_comparison["winner_2023_full_party"]
)


# Display the comparison.
display(
    state_winner_comparison
    .sort_values("state")
)


# Validation checks.
print(
    "States with missing official winners:",
    state_winner_comparison["winner_2023_full_party"].isna().sum()
)

print(
    "States where polling-unit dominant winner agrees with official winner:",
    state_winner_comparison["winner_agreement"].sum()
)

print(
    "States where they differ:",
    (~state_winner_comparison["winner_agreement"]).sum()
)
```

```python
# CELL 214 — Examine disagreements between polling-unit and official winners

# Identify states where the dominant polling-unit winner differs
# from the official 2023 full-party state winner.

winner_disagreements = (
    state_winner_comparison[
        ~state_winner_comparison["winner_agreement"]
    ]
    .sort_values(
        "dominant_pu_win_share",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    winner_disagreements[
        [
            "state",
            "dominant_pu_winner",
            "dominant_pu_win_share",
            "winner_2023_full_party",
            "winner_agreement"
        ]
    ]
)
```

```python
# CELL 215 — Compare polling-unit state vote shares with official 2023 results

# Standardize polling-unit state names before merging with the
# official state-level dataset.

pu_state_share_data = pu_state_votes.copy()

pu_state_share_data["state"] = (
    pu_state_share_data["state"]
    .astype(str)
    .str.strip()
    .str.title()
)

pu_state_share_data["state"] = (
    pu_state_share_data["state"]
    .replace({"Fct": "FCT"})
)


# Extract official 2023 party vote shares.
official_state_shares = (
    analysis_df[
        analysis_df["election_year"] == 2023
    ][
        [
            "state",
            "apc_2023_vote_share",
            "pdp_2023_vote_share",
            "lp_2023_vote_share",
            "nnpp_2023_vote_share"
        ]
    ]
    .drop_duplicates("state")
)

official_state_shares = official_state_shares.rename(
    columns={
        "apc_2023_vote_share": "official_apc_share",
        "pdp_2023_vote_share": "official_pdp_share",
        "lp_2023_vote_share": "official_lp_share",
        "nnpp_2023_vote_share": "official_nnpp_share"
    }
)


# Merge polling-unit aggregation with official state results.
state_share_comparison = (
    pu_state_share_data[
        [
            "state",
            "apc_vote_share",
            "pdp_vote_share",
            "lp_vote_share",
            "nnpp_vote_share"
        ]
    ]
    .merge(
        official_state_shares,
        on="state",
        how="left",
        validate="one_to_one"
    )
)


# Calculate percentage-point differences.
for party in ["apc", "pdp", "lp", "nnpp"]:
    state_share_comparison[
        f"{party}_share_difference_pp"
    ] = (
        state_share_comparison[f"{party}_vote_share"]
        - state_share_comparison[f"official_{party}_share"]
    )


print(
    "States compared:",
    len(state_share_comparison)
)

print(
    "States with missing official results:",
    state_share_comparison["official_apc_share"].isna().sum()
)

display(
    state_share_comparison
)
```

```python
# CELL 216 — Summarize polling-unit versus official state-level differences

difference_columns = [
    "apc_share_difference_pp",
    "pdp_share_difference_pp",
    "lp_share_difference_pp",
    "nnpp_share_difference_pp"
]

comparison_summary = pd.DataFrame({
    "party": ["APC", "PDP", "LP", "NNPP"],
    "mean_difference_pp": [
        state_share_comparison[col].mean()
        for col in difference_columns
    ],
    "mean_absolute_difference_pp": [
        state_share_comparison[col].abs().mean()
        for col in difference_columns
    ],
    "max_absolute_difference_pp": [
        state_share_comparison[col].abs().max()
        for col in difference_columns
    ]
})

display(
    comparison_summary
)
```

```python
# CELL 217 — Identify states with the largest vote-share discrepancies

# Calculate the largest absolute discrepancy across the four parties
# for each state.

state_share_comparison["largest_absolute_difference_pp"] = (
    state_share_comparison[
        difference_columns
    ]
    .abs()
    .max(axis=1)
)

state_share_comparison["party_with_largest_difference"] = (
    state_share_comparison[
        difference_columns
    ]
    .abs()
    .idxmax(axis=1)
    .str.replace(
        "_share_difference_pp",
        "",
        regex=False
    )
    .str.upper()
)

largest_discrepancies = (
    state_share_comparison
    .sort_values(
        "largest_absolute_difference_pp",
        ascending=False
    )
)

display(
    largest_discrepancies[
        [
            "state",
            "largest_absolute_difference_pp",
            "party_with_largest_difference"
        ]
        + difference_columns
    ]
    .head(15)
)
```

```python
# CELL 238 — Calculate election-to-election changes in key indicators
change_df = transition_df.copy()

grouped = change_df.groupby("state")

for variable in [
    "apc_vote_share",
    "pdp_vote_share",
    "other_party_vote_share",
    "turnout_rate",
    "enp_comparable",
    "comparable_margin_percentage_points"
]:
    change_df[f"{variable}_change"] = grouped[variable].diff()

change_df["previous_election"] = grouped["election_year"].shift(1)

display(
    change_df[
        [
            "state",
            "election_year",
            "previous_election",
            "apc_vote_share_change",
            "pdp_vote_share_change",
            "other_party_vote_share_change",
            "turnout_rate_change",
            "enp_comparable_change",
            "comparable_margin_percentage_points_change"
        ]
    ].head(12)
)
```

```python
# CELL 244 — Prepare the 2019–2023 regression dataset
regression_df = transition_changes[
    transition_changes["transition"] == "2019 → 2023"
].copy()

regression_variables = [
    "state",
    "apc_vote_share_change",
    "pdp_vote_share_change",
    "other_party_vote_share_change",
    "turnout_rate_change",
    "enp_comparable_change",
    "comparable_margin_percentage_points_change"
]

regression_df = regression_df[regression_variables].copy()

print("Regression observations:", len(regression_df))
print("States represented:", regression_df["state"].nunique())

display(regression_df.head(10))
```

```python
# CELL 246 — Model APC vote-share change using turnout, competition, and party-system change
import statsmodels.api as sm

X = regression_df[
    [
        "turnout_rate_change",
        "enp_comparable_change",
        "comparable_margin_percentage_points_change"
    ]
]

X = sm.add_constant(X)

y = regression_df["apc_vote_share_change"]

apc_model = sm.OLS(y, X).fit()

print(apc_model.summary())
```

```python
# CELL 247 — Extract the key APC regression statistics
apc_regression_results = pd.DataFrame({
    "variable": apc_model.params.index,
    "coefficient": apc_model.params.values,
    "p_value": apc_model.pvalues.values,
    "standard_error": apc_model.bse.values
})

apc_regression_results["significant_at_5pct"] = (
    apc_regression_results["p_value"] < 0.05
)

display(apc_regression_results.round(4))

print("R-squared:", round(apc_model.rsquared, 4))
print("Adjusted R-squared:", round(apc_model.rsquared_adj, 4))
```

```python
# CELL 248 — Model PDP vote-share change using turnout, competition, and party-system change
y_pdp = regression_df["pdp_vote_share_change"]

pdp_model = sm.OLS(y_pdp, X).fit()

print(pdp_model.summary())
```

```python
# CELL 249 — Extract the key PDP regression statistics
pdp_regression_results = pd.DataFrame({
    "variable": pdp_model.params.index,
    "coefficient": pdp_model.params.values,
    "p_value": pdp_model.pvalues.values,
    "standard_error": pdp_model.bse.values
})

pdp_regression_results["significant_at_5pct"] = (
    pdp_regression_results["p_value"] < 0.05
)

display(pdp_regression_results.round(4))

print("R-squared:", round(pdp_model.rsquared, 4))
print("Adjusted R-squared:", round(pdp_model.rsquared_adj, 4))
```

```python
# CELL 250 — Test the specific relationship between turnout change and party vote-share change
turnout_relationships = {}

for party in ["apc", "pdp", "other_party"]:
    x = regression_df[["turnout_rate_change"]]
    x = sm.add_constant(x)

    y = regression_df[f"{party}_vote_share_change"]

    model = sm.OLS(y, x).fit()

    turnout_relationships[party.upper()] = {
        "coefficient": model.params["turnout_rate_change"],
        "p_value": model.pvalues["turnout_rate_change"],
        "r_squared": model.rsquared
    }

turnout_relationships_df = (
    pd.DataFrame(turnout_relationships)
    .T
    .round(4)
)

display(turnout_relationships_df)
```

```python
# CELL 253 — Build the 2019–2023 state-level voting-behaviour clustering dataset

# Keep only the 2019 → 2023 election transition.
cluster_df = transition_changes[
    transition_changes["election_transition"] == "2019 → 2023"
].copy()

# Keep one observation per state.
cluster_variables = [
    "apc_vote_share_change",
    "pdp_vote_share_change",
    "enp_comparable_change"
]

cluster_data = (
    cluster_df[
        ["state"] + cluster_variables
    ]
    .drop_duplicates(subset="state")
    .copy()
)

print("Clustering period: 2019 → 2023")
print("States:", len(cluster_data))
print("Unique states:", cluster_data["state"].nunique())
print("Clustering variables:", len(cluster_variables))

display(cluster_data.head(10))
```

```python
# CELL 254 — Standardize variables for clustering
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_cluster = scaler.fit_transform(
    cluster_data[cluster_variables]
)

X_cluster_df = pd.DataFrame(
    X_cluster,
    columns=cluster_variables,
    index=cluster_data.index
)

display(
    X_cluster_df.describe()
    .round(3)
)
```

```python
# CELL 255 — Evaluate candidate numbers of clusters using silhouette scores
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

cluster_evaluation = []

for k in range(2, 7):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = kmeans.fit_predict(X_cluster)

    score = silhouette_score(
        X_cluster,
        labels
    )

    cluster_evaluation.append({
        "clusters": k,
        "silhouette_score": score
    })

cluster_evaluation_df = pd.DataFrame(
    cluster_evaluation
)

display(
    cluster_evaluation_df.round(4)
)
```

```python
# CELL 256 — Select the cluster count with the highest silhouette score
best_k = int(
    cluster_evaluation_df.loc[
        cluster_evaluation_df["silhouette_score"].idxmax(),
        "clusters"
    ]
)

best_score = cluster_evaluation_df.loc[
    cluster_evaluation_df["silhouette_score"].idxmax(),
    "silhouette_score"
]

print("Selected number of clusters:", best_k)
print("Best silhouette score:", round(best_score, 4))
```

```python
# CELL 257 — Fit the final K-means clustering model
final_kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=20
)

cluster_data["cluster"] = final_kmeans.fit_predict(
    X_cluster
)

cluster_data["cluster"] = (
    cluster_data["cluster"]
    .astype(int)
)

display(
    cluster_data[
        ["state", "cluster"] + cluster_variables
    ]
    .sort_values(["cluster", "state"])
)
```

```python
# CELL 258 — Profile the characteristics of each electoral cluster
cluster_profile = (
    cluster_data
    .groupby("cluster")[cluster_variables]
    .mean()
    .round(3)
)

cluster_profile["state_count"] = (
    cluster_data
    .groupby("cluster")
    .size()
)

display(cluster_profile)
```

```python
# CELL 259 — List states belonging to each electoral cluster
cluster_members = (
    cluster_data
    .groupby("cluster")["state"]
    .apply(list)
)

for cluster, states in cluster_members.items():
    print(f"\nCluster {cluster} ({len(states)} states):")
    print(", ".join(sorted(states)))
```

```python
# CELL 260 — Compare clusters using interpretable electoral indicators

cluster_interpretation = (
    cluster_data
    .groupby("cluster")
    .agg(
        states=("state", "count"),
        mean_apc_change=("apc_vote_share_change", "mean"),
        mean_pdp_change=("pdp_vote_share_change", "mean"),
        mean_enp_change=("enp_comparable_change", "mean")
    )
    .round(3)
)

display(cluster_interpretation)
```

```python
# CELL 262 — Check multicollinearity among regression predictors
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_data = regression_df[
    [
        "turnout_rate_change",
        "enp_comparable_change",
        "comparable_margin_percentage_points_change"
    ]
].copy()

vif_table = pd.DataFrame({
    "variable": vif_data.columns,
    "VIF": [
        variance_inflation_factor(
            vif_data.values,
            i
        )
        for i in range(vif_data.shape[1])
    ]
})

display(vif_table.round(3))
```

```python
# CELL 263 — Test normality of APC regression residuals
from scipy.stats import shapiro

shapiro_stat, shapiro_p = shapiro(
    apc_model.resid
)

print("Shapiro-Wilk statistic:", round(shapiro_stat, 4))
print("p-value:", round(shapiro_p, 4))
```

```python
# CELL 264 — Test for heteroskedasticity in the APC regression
from statsmodels.stats.diagnostic import het_breuschpagan

bp_test = het_breuschpagan(
    apc_model.resid,
    apc_model.model.exog
)

bp_results = pd.Series(
    bp_test,
    index=[
        "LM statistic",
        "LM p-value",
        "F statistic",
        "F p-value"
    ]
)

display(bp_results.round(4))
```

```python
# CELL 265 — Re-estimate APC regression using heteroskedasticity-robust standard errors
apc_model_robust = apc_model.get_robustcov_results(
    cov_type="HC3"
)

print(apc_model_robust.summary())
```

```python
# CELL 266 — Compare conventional and robust APC regression coefficients
robust_comparison = pd.DataFrame({
    "variable": apc_model.params.index,
    "coefficient_standard": apc_model.params.values,
    "p_value_standard": apc_model.pvalues.values,
    "coefficient_HC3": apc_model_robust.params,
    "p_value_HC3": apc_model_robust.pvalues
})

display(
    robust_comparison.round(4)
)
```

```python
# CELL 267 — Identify potentially influential states in the APC regression
influence = apc_model.get_influence()

influence_summary = influence.summary_frame()

influence_df = regression_df[
    ["state", "apc_vote_share_change"]
].copy()

influence_df["cooks_distance"] = (
    influence_summary["cooks_d"]
    .values
)

influence_df["leverage"] = (
    influence_summary["hat_diag"]
    .values
)

influence_df["studentized_residual"] = (
    influence_summary["student_resid"]
    .values
)

display(
    influence_df
    .sort_values("cooks_distance", ascending=False)
    .head(10)
)
```

```python
# CELL 268 — Conduct a leave-one-out robustness check for the most influential state
most_influential_state = (
    influence_df
    .sort_values("cooks_distance", ascending=False)
    .iloc[0]["state"]
)

robustness_subset = regression_df[
    regression_df["state"] != most_influential_state
].copy()

X_robust = robustness_subset[
    [
        "turnout_rate_change",
        "enp_comparable_change",
        "comparable_margin_percentage_points_change"
    ]
]

X_robust = sm.add_constant(X_robust)

y_robust = robustness_subset[
    "apc_vote_share_change"
]

apc_model_without_influential = sm.OLS(
    y_robust,
    X_robust
).fit()

print("State removed:", most_influential_state)
print(
    "Original R-squared:",
    round(apc_model.rsquared, 4)
)
print(
    "Robustness R-squared:",
    round(apc_model_without_influential.rsquared, 4)
)

display(
    pd.DataFrame({
        "original_coefficient": apc_model.params,
        "without_influential_state": apc_model_without_influential.params
    }).round(4)
)
```

```python
# CELL 269 — Test clustering stability across different random initializations
cluster_stability = []

for seed in [1, 10, 42, 100, 500]:
    model = KMeans(
        n_clusters=best_k,
        random_state=seed,
        n_init=20
    )

    labels = model.fit_predict(X_cluster)

    cluster_stability.append({
        "random_seed": seed,
        "inertia": model.inertia_,
        "silhouette_score": silhouette_score(
            X_cluster,
            labels
        )
    })

cluster_stability_df = pd.DataFrame(
    cluster_stability
)

display(
    cluster_stability_df.round(4)
)
```
